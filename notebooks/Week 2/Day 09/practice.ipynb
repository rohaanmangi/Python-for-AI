```python
def count_up_to(n):
    i = 1
    while i<= n:
        yield i
        i += 1

gen_func = count_up_to(5)
for number in gen_func:
    print(number)
```

```python
def shout(func):
    def wrapper(*args, **kwargs):
        result = func(*args, **kwargs)
        return result.upper()
    return wrapper

@shout
def greet(name):
    return f"hello {name}"

print(greet("rohaan"))
```
