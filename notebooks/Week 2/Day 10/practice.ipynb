```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum(self.marks)/len(self.marks)

    def grade(self):
        avg = self.average()
        if avg > 80:
            return "Grade: A"
        elif avg > 70:
            return "Grade: B"
        elif avg > 60:
            return "Grade: C"
        else:
            return "Grade: F"

fatima = Student("Fatima", [85, 92, 78])
print(fatima.name, fatima.average(), fatima.grade())
```

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
        name = "HBL"

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        self.balance -= amount

    def transfer(self, other_person_account, amount):
        if self.balance >= amount:
            self.withdraw(amount)
            other_person_account.deposit(amount)
        else:
            print("You dont have sufficient funds")


ali = BankAccount("Ali", 8000)
zara = BankAccount("Zara", 5000)

print("Balance of Ali", ali.balance)
print("Balance of Zara", zara.balance)

ali.transfer(zara, 5000)

print("Balance of Ali", ali.balance)
print("Balance of Zara", zara.balance)
```
